# Part 2: Interactive Chart 1 — Histogram - Survival counts

Create a bar‑style count chart of survival outcomes. Create a histogram including chart title, X-axis title,   and Y-axis title [ 2 marks].

```python
import plotly.express as px

survival_count = df['survived'].value_counts().reset_index()
survival_count.columns = ['Survived', 'Count']

fig = px.bar(
    survival_count,
    x='Survived',
    y='Count',
    title='Survival Count',
    labels={
        'Survived': 'Survival Status',
        'Count': 'Number of Passengers'
    }
)

fig.show()
```

**Write 2–3 sentences answering: [ 2 marks]**
Write 2 sentences describing what the chart shows.

This chart shows the number of passengers who survived and did not survive. The x-axis represents survival status, and the y-axis shows the total number of passengers in each group. It also displays the number of survivors and non-survivors as separate bars.
