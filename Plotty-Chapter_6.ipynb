```python
# This code creates a simple scatter plot using Matplotlib. 
# The plt.scatter() function displays the relationship between the x-values and y-values as individual points, 
# and the title and axis labels make the chart easier to understand.
x = [1, 2, 3]
y = [4, 5, 6]

plt.scatter(x, y)
plt.title("Matplotlib Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# A simple scatter plot using Seaborn. 
# The sns.scatterplot() function shows the relationship between the x- and y-values as individual points 
# plt.title() adds a title to the chart.

import seaborn as sns
import matplotlib.pyplot as plt
x = [1, 2, 3]
y = [4, 5, 6]

sns.scatterplot(x=x, y=y)
plt.title("Seaborn Scatter Plot")
plt.show()
```

```python
# This code creates a simple interactive scatter plot using Plotly Express. 
# Each point represents an (x,y)
# (x,y) pair, and fig.show() displays the chart in the notebook.

import plotly.express as px

fig = px.scatter(x=[1,2,3], y=[4,5,6])
fig.show()
```

The main difference is the library and the experience each one gives you. Matplotlib is the most basic and customizable, Seaborn is built on top of Matplotlib and makes plots look nicer with less code, and Plotly creates interactive charts that you can zoom, hover, and explore.

In your examples
Matplotlib: plt.scatter(x, y) makes a static scatter plot, and you manually add the title and axis labels.

Seaborn: sns.scatterplot(x=x, y=y) also makes a static scatter plot, but with a cleaner default style.

Plotly: px.scatter(x=[1,2,3], y=[4,5,6]) makes an interactive scatter plot that opens with fig.show().

# Customizing / updating Axes and Titles
Customization in plotting means adjusting the default appearance of a chart so that it is clearer, more visually appealing, and better suited to your context. In Plotly, you can customize things like:

Axes: change the axis titles, range, grid, and tick labels.

Titles and labels: set or change the chart title, axis labels, and legend text.

Style: change colors, fonts, spacing, and layout.

```python
data = {"Category":['A','B','C','D','E'], "Value":[20,35,30,50,40]}
df=pd.DataFrame(data)
fig = px.bar(df,x='Category',y='Value', title="Category vs Value")
fig.show()
```

```python
#This example shows how to customize the axes labels and title of a Plotly bar chart. 
# The data is first stored in a pandas DataFrame with a Category column and a Value column, and then px.bar() creates a basic bar plot. The update_layout() method is used to set the x‑axis label to "Category", the y‑axis label to "Values", 
# the overall title to "Customized Bar Chart", 
# before displaying the interactive chart with fig.show().
import plotly.express as px
import pandas as pd 
data = {"Category":['A','B','C','D','E'], "Value":[20,35,30,50,40]}
df=pd.DataFrame(data)
fig = px.bar(df,x='Category',y='Value')
fig.update_layout(xaxis_title="Category",yaxis_title="Values",title="Customized Bar Chart")
fig.show()
```

# Changing colors and showing markers
This  following example shows how to customize the color and marker style of a Plotly line chart. The markers=True argument adds circles at each data point on the line, and color_discrete_sequence=["green"] sets the line color to green. The chart is then displayed interactively with fig.show().

```python
# Changing colors and showing markers
fig = px.line( x=[1, 2, 3, 4], y=[10, 15, 13, 17], markers=True, color_discrete_sequence=["green"])
fig.show()
```

```python
# Build a line chart where markers are shown at each data point using markers=True. 
# The line color is then changed to blue with line=dict(color="blue"), 
# the markers are customized separately using 
# marker=dict(color="orange", size=10, symbol='star'), make the points large orange star symbols.

fig = px.line( x=[1,2,3,4], y=[10,15,13,17], markers=True)
fig.update_traces( line=dict(color="blue"), marker=dict(color="orange", size=10,symbol='star'))
fig.show()
```

```python
# Create a basic scatter plot using px.scatter() with simple x and y values. 
# Then fig.update_traces(marker=dict(color="orange", size=12)) 
# Customize the markers so that all points are orange and larger than the default size. 
fig = px.scatter(x=[1,2,3,4], y=[10,15,13,17])
fig.update_traces( marker=dict(color="orange", size=12))
fig.show()
```

```python
# The first fig.show() displays the chart with default colors. 
# Update_traces(marker=dict(colors=["purple","lightgreen","orange"])) is used to manually assign custom colors to each slice, 
# The second fig.show() displays the re‑styled pie chart with the new color scheme.
import plotly.express as px
fig = px.pie( names=["A","B","C"], values=[10,20,30])
fig.show()
fig.update_traces( marker=dict(colors=["purple","lightgreen","orange"]))
fig.show()
```

```python
# Define a theme at creation time
import plotly.express as px
fig = px.pie( names=["A","B","C","D"], values=[20,40,15,25], color_discrete_sequence=["red","blue","green","orange"])
fig.show()
```

# Annotations in Plotly

Annotations are small text labels or callouts you can add to a plot to highlight specific points, draw attention to important values, or add short explanations directly on the chart. In Plotly, you add them with the add_annotation() method, which lets you control:

Text content,

Position on the x‑ and y‑axes,

Arrow style (whether to show an arrow and how it points),

and fine‑tune offset so the label does not cover the data.

```python
# Adding Annotations

import plotly.express as px
import pandas as pd
data = {"Category":['A','B','C','D','E'],"Value":[20,35,30,50,40]}
df=pd.DataFrame(data)
fig=px.bar(df,x='Category',y='Value',title='Interactive Bar Chart with Annotations')

# Adds a text annotation pointing to the bar at Category `'B'` (x=1, value 35). 
# The arrow connects the label to the bar, and the small vertical offset (`ay=-40`) moves the text slightly above the bar so it remains readable. 

fig.add_annotation(x=1,y=35, text="Peak Value",showarrow=True,arrowhead=2,ax=0,ay=-40)
fig.show()
```

# Visualizing Geospatial Data
Plotly also supports geospatial visualizations, allowing you to map data points across the globe. This is particularly useful for data such as location-based analytics, geographic distributions and spatial relationships.
Let's see scatter map. You will need latitude and longitude coordinates for data points.

```python
import plotly.express as px
import pandas as pd
data = {"City":['New York','Los Angeles','Chicago','Houston','Phoenix'], "Latitude":[40.7128,34.0522,41.8781,29.7604,33.4484], "Longitude":[-74.0060,-118.2437,-87.6298,-95.3698,-112.0740]}
df=pd.DataFrame(data)
#Create a scatter map
fig = px.scatter_geo(df,lat="Latitude",lon="Longitude",text="City",title="Cities in the USA")
#Show the map
fig.show()
```

# Choropleth Maps 
A choropleth map allows you to visualize values across a geographic area, with different colors representing different values.

```python
import plotly.express as px
import pandas as pd
data = {"State": ['California','Texas','Florida','NewYork','Pennsylvania'], "ISO-code":['CA','TX','FL','NY','PA'], "Population":[39512223,28995881,21477737,19453561,12801989]}
df=pd.DataFrame(data)
#Create a choropleth map
fig = px.choropleth(df,locations="ISO-code",locationmode="USA-states",color="Population",color_continuous_scale="Viridis",
title="USA State Population",scope="usa", hover_name="State",) #scope="usa"
#Show the map
fig.show()
```

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 140, 210, 190]
})

# produces an interactive line chart with the following properties
# Monthly sales data for 6 months, plotted as a line with markers with formatted axis labels and title.
# Add Code Here .......

fig = px.line(df, x="Month", y="Sales", title="Monthly Sales Trend", markers=True)

fig.update_layout(
    template="plotly_white",
    title_font_size=22,
    xaxis_title="Month",
    yaxis_title="Sales ($)",
    showlegend=False
)
# Custom line color and marker size applied to the trace.
# Add Code Here  ......
fig.update_traces(line_color="darkblue", marker_size=10)

# Add Two text annotations: 
# First  at the highest sales point labeled "Highest sales" 
# Second  at the dip labeled "Sales dip".
# Add Code Here 
fig.add_annotation(
    x="May",
    y=210,
    text="Highest sales",
    showarrow=True,
    arrowhead=2,
    ax=0,
    ay=-40
)

fig.add_annotation(
    x="Apr",
    y=140,
    text="Sales dip",
    showarrow=True,
    arrowhead=2,
    ax=30,
    ay=40
)

fig.show()
```
